on ne veut récupérer de valeurs_trimestrielles.csv que les lignes qui nous interessent

```python
import pandas as pd

# Charger le fichier CSV d'origine
df = pd.read_csv("../data/chomage/valeurs_trimestrielles.csv", sep=';', encoding='utf-8')

# Filtrer les lignes dont le libellé (1ère colonne) n'est pas "Codes"
df_filtré = df[df.iloc[:, 0] != "Codes"]

# Exporter le résultat
df_filtré.to_csv("valeurs_trimestrielles_sans_codes.csv", sep=';', index=False, encoding='utf-8')
```
