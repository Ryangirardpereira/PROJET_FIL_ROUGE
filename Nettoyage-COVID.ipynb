```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker

# Charger les données
df = pd.read_csv("data/COVID/owid-covid-data.csv")
df_fr = df[df["location"] == "France"]
df_fr = df_fr[["date", "total_cases"]].dropna()
df_fr["date"] = pd.to_datetime(df_fr["date"])

# Créer la figure et l'axe
fig, ax = plt.subplots(figsize=(12, 6))

# Tracer la courbe sur l'axe
ax.plot(df_fr["date"], df_fr["total_cases"], color='darkred', linewidth=2)

# Personnaliser les axes
ax.set_title("Évolution du nombre total de cas Covid en France", fontsize=14)
ax.set_xlabel("Date")
ax.set_ylabel("Nombre total de cas")
ax.yaxis.set_major_formatter(ticker.FuncFormatter(lambda x, p: format(int(x), ',')))

# Grille + affichage
ax.grid(True)
plt.tight_layout()
plt.show()
```
