Bout de code qui permet de convertir des .dbf en .csv.
On a des .csv pour les années 2018 à 2023, mais les années 2007 à 2017 sont en .dbf.
On va donc tout convertir en .csv pour traiter toutes les années de la meme manière.

```python
from dbfread import DBF
import csv

def convertir_dbf_vers_csv(chemin_dbf, chemin_csv):
    """
    Convertit un fichier .dbf en fichier .csv.

    Args:
        chemin_dbf (str): Le chemin du fichier .dbf d'entrée.
        chemin_csv (str): Le chemin du fichier .csv de sortie.
    """
    try:
        table = DBF(chemin_dbf)
        with open(chemin_csv, 'w', newline='', encoding='utf-8') as fichier_csv:
            writer = csv.writer(fichier_csv)
            writer.writerow(table.field_names)
            for record in table:
                writer.writerow(list(record.values()))
        print(f"Le fichier '{chemin_dbf}' a été converti avec succès en '{chemin_csv}'.")
    except FileNotFoundError:
        print(f"Erreur : Le fichier '{chemin_dbf}' n'a pas été trouvé.")
    except Exception as e:
        print(f"Une erreur s'est produite lors de la conversion : {e}")

if __name__ == "__main__":
    nom_fichier_dbf = input("Veuillez entrer le nom du fichier .dbf à convertir : ")
    nom_fichier_csv = input("Veuillez entrer le nom du fichier .csv de sortie : ")
    convertir_dbf_vers_csv(nom_fichier_dbf, nom_fichier_csv)
```
